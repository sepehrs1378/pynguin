```python
import os
import json
```

```python
BASE_PATH = "../"

def is_json_serializable(line):
    """Check if a line can be converted to JSON."""
    try:
        json.loads(line)
        return True
    except (ValueError, TypeError):
        return False

def process_files(source_folder, dest_folder):
    """Process all files in source folder and write JSON lines to destination folder."""
    
    source_folder = BASE_PATH + source_folder
    dest_folder = BASE_PATH + dest_folder
    
    os.makedirs(dest_folder, exist_ok=True)
    for filename in os.listdir(source_folder):
        source_path = os.path.join(source_folder, filename)
        dest_path = os.path.join(dest_folder, filename)
        
        # Skip directories
        if os.path.isdir(source_path):
            continue
            
        try:
            with open(source_path, 'r', encoding='utf-8') as src_file, \
                 open(dest_path, 'w', encoding='utf-8') as dest_file:
                
                for line in src_file:
                    line = line.strip()
                    if line and is_json_serializable(line):
                        dest_file.write(line + '\n')
                        
            print(f"Processed: {filename}")
            
        except Exception as e:
            print(f"Error processing {filename}: {str(e)}")
        
```

```python
source_folder = 'results/raw/'
dest_folder = 'results/refined/'  # Change this to your desired output folder
process_files(source_folder, dest_folder)
print('Done')
```
