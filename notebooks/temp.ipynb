```python
s = """54 & 117 & 72 & 136 & 165 & 76 \\
88 & 37 & 253 & 110 & 81 & 86 \\
115 & 63 & 345 & 244 & 120 & 111 \\
116 & 64 & 222 & 251 & 115 & 96 \\
51 & 58 & 233 & 155 & 138 & 108 \\
82 & 120 & 161 & 67 & 112 & 81 \\
69 & 130 & 49 & 209 & 210 & 125 \\
37 & 124 & 226 & 397 & 144 & 84 \\
92 & 98 & 108 & 79 & 106 & 56 \\
169 & 86 & 144 & 169 & 106 & 128 \\
166 & 89 & 432 & 287 & 85 & 100 \\
105 & 92 & 193 & 109 & 116 & 143 \\
99 & 121 & 106 & 126 & 82 & 109 \\
101 & 46 & 113 & 85 & 111 & 97 \\
110 & 94 & 119 & 167 & 82 & 139 \\"""
```

```python
lines = s.split("\n")
lines = [l.replace("\\", "").replace(" ", "") for l in lines]
numbers_list = [list(map(int, l.split("&"))) for l in lines]

runs = {"a": {}, "d": {}, "e": {}}
for i, numbers in enumerate(numbers_list):
    runs["a"][i+1] = numbers[0]
    runs["a"][i+16] = numbers[1]
    runs["d"][i+1] = numbers[2]
    runs["d"][i+16] = numbers[3]
    runs["e"][i+1] = numbers[4]
    runs["e"][i+16] = numbers[5]

for i in range(1, 16):
    print(f"{i} & {runs['a'][i]} & {runs['d'][i]} & {runs['e'][i]} & {i+15} & {runs['a'][i+15]} & {runs['d'][i+15]} & {runs['e'][i+15]} \\\\")
```
